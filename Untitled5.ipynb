```python
from PIL import Image

img = Image.open("gato.jpg")
img = img.convert("RGB")
largura, altura = img.size
pixels = list(img.getdata())
```

```python
def binarizar(pixels, limiar=128):
  resultado = []
  for r, g, b in pixels:
    cinza = (r + g + b) // 3
    if cinza >= limiar:
      resultado.append(255)
    else:
      resultado.append(0)
  return resultado
```

```python
binarios = binarizar(pixels)
img_bin = Image.new("L", (largura, altura))
img_bin.putdata(binarios)
img_bin.save("gato_binario.jpg")
img_bin
```
